## Prompt
today I’m continuing with the Superstore Sales dataset from Day 1. Please guide me line by line to clean the data in my Jupyter notebook. For each line, explain its purpose (e.g., handling missing values, checking data types, removing outliers). Wait for my confirmation that the code runs successfully before giving the next line.
```python
print(df.isnull().sum())
```

```python
df['Postal Code'] = df['Postal Code'].fillna('Unknown')
print(df.isnull().sum())
```

```python
print(df.dtypes)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)
print(df[['Order Date', 'Ship Date']].dtypes)
```

```python
df['Order Date Shamsi'] = df['Order Date'].apply(lambda x: jdatetime.date.fromgregorian(date=x).strftime('%Y/%m/%d'))
df['Ship Date Shamsi'] = df['Ship Date'].apply(lambda x: jdatetime.date.fromgregorian(date=x).strftime('%Y/%m/%d'))
print(df[['Order Date', 'Order Date Shamsi', 'Ship Date', 'Ship Date Shamsi']].head())
```

```python
sales_99th_percentile = df['Sales'].quantile(0.99)
print(f"99th percentile of Sales: {sales_99th_percentile}")
```

```python
df = df[df['Sales'] <= sales_99th_percentile]
print("Shape after removing outliers:", df.shape)
```

```python
print("Null counts:\n", df.isnull().sum())
print("\nData types:\n", df.dtypes)
print("\nSample of cleaned data:\n", df[['Order Date', 'Order Date Shamsi', 'Ship Date', 'Ship Date Shamsi', 'Sales']].head())
```

```python
df = df.sort_values('Order Date').reset_index(drop=True)
```

```python
df['lag_1_sales'] = df['Sales'].shift(1)
```

```python
df = pd.get_dummies(df, columns = ['Category'], prefix = 'cat')
```

```python
df
```

```python
df['order_month'] = df['Order Date'].dt.month
```

```python
print(df['order_month'].unique())
```
